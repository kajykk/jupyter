```python
import numpy as np

# 混淆矩阵
cm = np.array([[10, 3, 5],
               [1, 15, 6],
               [2, 4, 20]])

# 计算每个类别的精确率、召回率、F1分数和支持度
precision = cm.diagonal() / cm.sum(axis=0)
recall = cm.diagonal() / cm.sum(axis=1)
f1 = 2 * (precision * recall) / (precision + recall)
support = cm.sum(axis=1)

# 计算总体准确率
accuracy = cm.diagonal().sum() / cm.sum()

# 计算宏观平均
macro_precision = np.mean(precision)
macro_recall = np.mean(recall)
macro_f1 = np.mean(f1)

# 计算加权平均
weights = support / support.sum()
weighted_precision = np.sum(precision * weights)
weighted_recall = np.sum(recall * weights)
weighted_f1 = np.sum(f1 * weights)

# 打印结果
print("              precision    recall     f1-score    support")
for i, label in enumerate(["猫", "狗", "猪"]):
    print(f"{label:<10}   {precision[i]:.1f}       {recall[i]:.1f}      {f1[i]:.1f}       {support[i]:.0f}")

print("\naccuracy    ", accuracy)
print("macro avg   ", macro_precision, macro_recall, macro_f1)
print("weighted avg", weighted_precision, weighted_recall, weighted_f1)

def calculate_segmentation_metrics(gold_standard, segmentation_result, dictionary):
    """
    计算分词指标

    参数：
    gold_standard -- 标准答案的分词列表
    segmentation_result -- 分词结果列表
    dictionary -- 词典列表

    返回：
    metrics -- 包含各项指标的字典
    """
    # 初始化变量
    correct_segments = 0  # 正确切分的词数
    predicted_segments = len(segmentation_result)  # 预测切分的词数
    actual_segments = len(gold_standard)  # 实际词数
    iv_correct = 0  # 正确切分的词典内词数
    iv_actual = 0  # 实际词典内词数
    oov_correct = 0  # 正确切分的词典外词数
    oov_actual = 0  # 实际词典外词数

    # 计算各个指标需要的值
    for word in gold_standard:
        if word in dictionary:
            iv_actual += 1
        else:
            oov_actual += 1

    for predicted_word in segmentation_result:
        if predicted_word in gold_standard:
            correct_segments += 1
            if predicted_word in dictionary:
                iv_correct += 1
            else:
                oov_correct += 1

    # 计算 Precision、Recall 和 F1 Score
    precision = correct_segments / predicted_segments if predicted_segments != 0 else 0
    recall = correct_segments / actual_segments if actual_segments != 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) != 0 else 0

    # 计算 IV Recall
    iv_recall = iv_correct / iv_actual if iv_actual != 0 else 0

    # 计算 OOV Recall
    oov_recall = oov_correct / oov_actual if oov_actual != 0 else 0

    # 创建包含结果的字典
    metrics = {
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "IV Recall": iv_recall,
        "OOV Recall": oov_recall
    }

    return metrics


# 定义标准答案、分词结果和词典
gold_standard = ["我家", "门前", "的", "小河", "很", "难", "过"]
segmentation_result = ["我家", "门前", "的", "小河", "很难过"]
dictionary = ["我家", "门前", "的", "小河", "难过"]

# 计算指标
metrics = calculate_segmentation_metrics(gold_standard, segmentation_result, dictionary)

# 打印结果
print("Precision:", metrics["Precisionyghbv"])
print("Recall:", metrics["Recall"])
print("F1 Score:", metrics["F1 Score"])
print("IV Recall:", metrics["IV Recall"])
print("OOV Recall:", metrics["OOV Recall"])
```
